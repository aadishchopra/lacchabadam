```python
#%pip install pandas
#%pip install transformers torch accelerate
#%pip install huggingface_hub
#%pip install ipywidgets
import pandas
import matplotlib.pyplot as plt
import flask
import sklearn
import ipywidgets as widgets


from transformers import AutoTokenizer, AutoModelForCausalLM
import torch


# Optionally login if using gated models
from huggingface_hub import notebook_login
# notebook_login()  # Uncomment and run once if model access requires auth

# Try smallest public model first for reliability
model_name = "gpt2"  # small, public, and works without approval

try:
    print(f"Loading tokenizer for {model_name}...")
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    print(f"Loading model {model_name}...")
    model = AutoModelForCausalLM.from_pretrained(model_name)
    print("Model loaded successfully.")
except Exception as e:
    print("Primary model load failed:", e)
    print("Falling back to tiny model to reduce memory and CPU usage.")
    model_name = "sshleifer/tiny-gpt2"
    try:
        tokenizer = AutoTokenizer.from_pretrained(model_name)
        model = AutoModelForCausalLM.from_pretrained(model_name)
        print("Fallback model loaded successfully.")
    except Exception as e2:
        print("Fallback model load failed:", e2)
        raise RuntimeError("Cannot load any model - check network or environment.")

# Example prompt
prompt = "Explain how APIs work in simple terms."

# Tokenize and generate
inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
outputs = model.generate(inputs["input_ids"], max_length=200, temperature=0.7)
response = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(response)
```
